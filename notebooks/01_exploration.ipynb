```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
DATA_DIR = Path("../data/raw/real_vs_fake/real-vs-fake")
TRAIN_REAL = DATA_DIR / "train/real"
TRAIN_FAKE = DATA_DIR / "train/fake"
```

```python
print(f"Train Real: {len(list(TRAIN_REAL.glob('*.jpg')))}")
print(f"Train Fake: {len(list(TRAIN_FAKE.glob('*.jpg')))}")
```

```python
fig, axes = plt.subplots(2, 5, figsize=(15, 6))

real_images = list(TRAIN_REAL.glob('*.jpg'))[:5]
fake_images = list(TRAIN_FAKE.glob('*.jpg'))[:5]

for i, img_path in enumerate(real_images):
    img = cv2.imread(str(img_path))
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    axes[0, i].imshow(img)
    axes[0, i].set_title('REAL', color='green', fontweight='bold')
    axes[0, i].axis('off')

for i, img_path in enumerate(fake_images):
    img = cv2.imread(str(img_path))
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    axes[1, i].imshow(img)
    axes[1, i].set_title('FAKE', color='red', fontweight='bold')
    axes[1, i].axis('off')

plt.suptitle('Sample: Real vs Fake Faces', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('../experiments/sample_visualization.png', dpi=150)
plt.show()
```
